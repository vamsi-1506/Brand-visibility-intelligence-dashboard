```python
import pandas as pd
import re

# ============================================================
# 1. LOAD THE DATASET
# ============================================================



print("Shape:", df_csv.shape)
print("\nColumns:")
print(df_csv.columns)


# ============================================================
# 2. LIST OF KNOWN BRANDS
# ============================================================

brands = [
    'Apple',
    'Samsung',
    'Sony',
    'LG',
    'HP',
    'Dell',
    'Lenovo',
    'Asus',
    'Acer',
    'Microsoft',
    'Google',
    'OnePlus',
    'Xiaomi',
    'Realme',
    'Oppo',
    'Vivo',
    'Nokia',
    'Canon',
    'Nikon',
    'Bose',
    'JBL',
    'Philips',
    'Whirlpool',
    'Haier',
    'Bosch',
    'Hotpoint',
    'Insignia',
    'Frigidaire',
    'GE',
    'TCL',
    'Vizio',
    'Puma',
    'Nike',
    'Adidas',
    'Boat',
    'Bajaj',
    'IFB',
    'Godrej',
    'Panasonic',
    'Motorola',
    'Nothing'
]


# ============================================================
# 3. FUNCTION TO EXTRACT BRAND
# ============================================================

def extract_brand(title):

    # If title is missing
    if pd.isna(title):
        return 'Unknown'

    title = str(title).strip()

    # If title is empty
    if not title:
        return 'Unknown'

    # --------------------------------------------------------
    # Step 1: Check the first word
    # --------------------------------------------------------

    first_word = title.split()[0]

    for brand in brands:

        if first_word.lower() == brand.lower():
            return brand

    # --------------------------------------------------------
    # Step 2: Search the complete title
    # --------------------------------------------------------

    for brand in brands:

        pattern = r'\b' + re.escape(brand) + r'\b'

        if re.search(pattern, title, flags=re.IGNORECASE):
            return brand

    # --------------------------------------------------------
    # Step 3: If no brand is found
    # --------------------------------------------------------

    return 'Unknown'


# ============================================================
# 4. CREATE BRAND COLUMN
# ============================================================

df_csv['brand'] = df_csv['title'].apply(extract_brand)


# ============================================================
# 5. CHECK THE RESULT
# ============================================================

print("\nFirst 30 records:")
print(df_csv[['title', 'brand']].head(30))


# ============================================================
# 6. CHECK BRAND COUNTS
# ============================================================

print("\nBrand Counts:")
print(df_csv['brand'].value_counts())


# ============================================================
# 7. CHECK UNKNOWN BRANDS
# ============================================================

unknown_count = (df_csv['brand'] == 'Unknown').sum()

print("\nNumber of Unknown brands:", unknown_count)

print(
    "Percentage of Unknown brands:",
    round((unknown_count / len(df_csv)) * 100, 2),
    "%"
)


# ============================================================
# 8. DISPLAY TITLES WHERE BRAND IS UNKNOWN
# ============================================================

print("\nTitles with Unknown brand:")

print(
    df_csv.loc[
        df_csv['brand'] == 'Unknown',
        ['title']
    ].head(50)
)


# ============================================================
# 9. FINAL DATASET INFORMATION
# ============================================================

print("\nFinal Shape:", df_csv.shape)

print("\nFinal Columns:")
print(df_csv.columns)

print("\nMissing Values:")
print(df_csv.isnull().sum())


```
